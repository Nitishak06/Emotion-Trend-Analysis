```python
import requests

url = "http://127.0.0.1:8080/analyze"
test_message = {"text": "I am  feeling low"}

try:
    response = requests.post(url, json=test_message)
    print("--- REAL AI RESULT ---")
    # This will show the topic, confidence, and message
    print(response.json())
except Exception as e:
    print(f"Error connecting: {e}")
```
